```python
mask = (
    (master_results["Experiment"] == "PlantVillage → PlantVillage")
    &
    (master_results["Model"] == "Baseline CNN")
)

master_results.loc[mask, "Macro_F1"] = 0.937383
master_results.loc[mask, "Weighted_F1"] = 0.951644

# Use the exact accuracy from the classification report too
master_results.loc[mask, "Accuracy"] = 0.951633

master_results.to_csv(
    "../report/master_model_results.csv",
    index=False
)

master_results
```

```python
import matplotlib.pyplot as plt
import numpy as np

plot_df = master_results.pivot(
    index="Experiment",
    columns="Model",
    values="Accuracy"
) * 100

# Put experiments in a logical order
plot_df = plot_df.reindex([
    "PlantVillage → PlantVillage",
    "PlantDoc → PlantDoc",
    "PlantVillage → PlantDoc"
])

ax = plot_df.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.ylabel("Accuracy (%)")
plt.xlabel("")
plt.title("Model Performance Across Within-Dataset and Cross-Dataset Evaluation")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(title="Model")
plt.grid(axis="y", alpha=0.3)

# Put values above bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3,
        fontsize=9
    )

plt.tight_layout()

plt.savefig(
    "../report/final_model_accuracy_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Macro F1 across all three experiments
import matplotlib.pyplot as plt

macro_plot_df = master_results.pivot(
    index="Experiment",
    columns="Model",
    values="Macro_F1"
) * 100

macro_plot_df = macro_plot_df.reindex([
    "PlantVillage → PlantVillage",
    "PlantDoc → PlantDoc",
    "PlantVillage → PlantDoc"
])

ax = macro_plot_df.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.ylabel("Macro F1-score (%)")
plt.xlabel("")
plt.title("Macro F1 Performance Across Experimental Settings")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(title="Model")
plt.grid(axis="y", alpha=0.3)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3,
        fontsize=9
    )

plt.tight_layout()

plt.savefig(
    "../report/final_model_macro_f1_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#domine shift table
domain_shift = pd.DataFrame({
    "Model": [
        "Baseline CNN",
        "MobileNetV2",
        "EfficientNetB0"
    ],

    "PlantVillage_Accuracy": [
        0.951633,
        0.950528,
        0.955807
    ],

    "CrossDataset_Accuracy": [
        0.055085,
        0.275424,
        0.313559
    ]
})

domain_shift["Accuracy_Drop"] = (
    domain_shift["PlantVillage_Accuracy"]
    - domain_shift["CrossDataset_Accuracy"]
)

domain_shift["Relative_Accuracy_Retention"] = (
    domain_shift["CrossDataset_Accuracy"]
    / domain_shift["PlantVillage_Accuracy"]
)

domain_shift
```

```python
domain_shift_display = domain_shift.copy()

for col in [
    "PlantVillage_Accuracy",
    "CrossDataset_Accuracy",
    "Accuracy_Drop",
    "Relative_Accuracy_Retention"
]:
    domain_shift_display[col] = (
        domain_shift_display[col] * 100
    ).round(2)

domain_shift_display
```

```python
domain_shift_display.to_csv(
    "../report/domain_shift_summary.csv",
    index=False
)
```
